# Day 6 — Savings Goal Tracker Agent Demo

This notebook demonstrates the **Savings Goal Tracker Agent**, an AI agent that assists users in setting, tracking, and reaching financial savings goals.

### Agentic Architecture:
- **USER** $\rightarrow$ **AGENT PLAN** $\rightarrow$ **TOOL CALL** $\rightarrow$ **TOOL RESULT** $\rightarrow$ **MEMORY** $\rightarrow$ **AGENT DECISION** $\rightarrow$ **FINAL ANSWER**
- **Deterministic Calculations**: Pure Python math for targets, total saved, remaining amount, completion percentage, time metrics, and status (`NOT_STARTED`, `BEHIND`, `ON_TRACK`, `GOAL_REACHED`).
- **Cross-Turn Memory**: Persistent `SavingsMemory` retaining goal parameters and multiple savings entries.
